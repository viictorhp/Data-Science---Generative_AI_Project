```python
from typing import TypedDict, Literal
```

```python
class EstadoAgente(TypedDict):
  pregunta: str # La pregunta del usuario
  respuesta: str # La respuesta generada por el modelo
  procesado: int # Si la pregunta ya ha sido procesada o no
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
MODEL = "gemini-2.5-flash-lite"
```

```python
llm = ChatGoogleGenerativeAI(
    model = MODEL,
    temperature = 0.5,
    google_api_key = API_KEY
)
```

```python
from langgraph.graph import StateGraph, START, END
```

```python
# 1. Crear el grafo con el objeto estado definido
grafo = StateGraph(EstadoAgente)
```

```python
def nodo_preprocesar(estado: EstadoAgente) -> dict:
  """
  Este nodo limpia y prepara la pregunta antes de envirarla al LLM.
  Retorna solo los campos que quiere modificar
  """
  pregunta_limpia = "¿Cual es la capital de España?"
  print(f"[Nodo Preprocesar] - Pregunta Original: '{estado["pregunta"]}'")
  print(f"[Nodo Preprocesar] - Pregunta nueva: '{pregunta_limpia}'")
  print(f"[Nodo Numero]:'{estado["procesado"]+1}'")

  return{
      "pregunta": pregunta_limpia,
      "procesado": 1
  }
```

```python
def nodo_responder(estado: EstadoAgente) -> dict:
  """
  Este nodo envia la pregunta al LLM y guarda la respuesta del modelo en el estado
  """
  respuesta = llm.invoke(estado["pregunta"])

  print(f"[Nodo Numero]:'{estado["procesado"]+1}'")

  return{
      "respuesta": respuesta.content,
      "procesado": 2
  }
```

```python
# 2. Agregar nodos
grafo.add_node("preprocesar", nodo_preprocesar)
grafo.add_node("responder", nodo_responder)
```

```python
grafo.add_edge(START, "preprocesar")

grafo.add_edge("preprocesar", "responder")

grafo.add_edge("responder", END)
```

```python
app = grafo.compile()
```

# Condicionales

```python
class EstadoAsistente(TypedDict):
  pregunta: str
  tipo: str
  respuesta: str
  nodo_usado: str
```

```python
def nodo_clasificar(estado: EstadoAsistente) -> dict:
  """
  Clasifica la pregunta del usuario en una de tres categorias
  """
  prompt = f""" Clasifica la siguiente pregunta en UNA de estas tres categorias:
  - matematicas (si involucra cualquier calculo, algebra, estadistica, logica)
  - codigo (si pide escribir, debuggear o explicar codigo)
  - cultura (otra pregunta de contexto general)
  Responde SOLO con una de estas tres palabras: matematicas, codigo, cultura
  Pregunta: {estado["pregunta"]} """

  respuesta = llm.invoke(prompt)
  tipo = respuesta.content.strip().lower()

  print(f"[Pregunta original]: '{estado['pregunta']}'" )
  print(f"[Clasificacion calculada]: '{tipo}'" )

  return {
      "tipo": tipo
  }
```

```python
def nodo_matematicas(estado: EstadoAsistente) -> dict:
    """
    Resuelve preguntas matemáticas paso a paso.
    """
    prompt = f"""Eres un tutor de matemáticas. Resuelve la siguiente pregunta mostrando
cada paso claramente. Usa notación clara.

Pregunta: {estado['pregunta']}"""

    respuesta = llm.invoke(prompt)
    print("[Nodo Matemáticas] Procesando...")
    return {
        "respuesta": respuesta.content,
        "nodo_usado": "matematicas"
    }


def nodo_cultura(estado: EstadoAsistente) -> dict:
    """
    Responde preguntas de cultura general de forma concisa.
    """
    prompt = f"""Eres un asistente enciclopédico. Responde de forma concisa y clara.

Pregunta: {estado['pregunta']}"""

    respuesta = llm.invoke(prompt)
    print("[Nodo Cultura] Procesando...")
    return {
        "respuesta": respuesta.content,
        "nodo_usado": "cultura"
    }


def nodo_codigo(estado: EstadoAsistente) -> dict:
    """
    Ayuda con preguntas de programación, siempre con ejemplo de código.
    """
    prompt = f"""Eres un experto en programación. Responde siempre incluyendo un ejemplo
de código funcional. Usa bloques de código con el lenguaje apropiado.

Pregunta: {estado['pregunta']}"""

    respuesta = llm.invoke(prompt)
    print("[Nodo Código] Procesando...")
    return {
        "respuesta": respuesta.content,
        "nodo_usado": "codigo"
    }
```

```python
def router(estado: EstadoAsistente) -> Literal["matematicas","cultura","codigo"]:
  """ Funcion router que lee el tipo del estado y retorna el nombre del nodo destino.
  El valor retornado debe corresponder a un nombre de nodo registrado en el grafo
  """

  tipo = estado["tipo"]

  print(f"[Router] Dirigiendo al nodo: '{tipo}'")

  return tipo # "matematicas","cultura","codigo"
```

```python
grafo_condicional = StateGraph(EstadoAsistente)
```

```python
grafo_condicional.add_node("clasificar", nodo_clasificar)
grafo_condicional.add_node("matematicas", nodo_matematicas)
grafo_condicional.add_node("cultura", nodo_cultura)
grafo_condicional.add_node("codigo", nodo_codigo)
```

```python
grafo_condicional.add_edge(START, "clasificar")
```

```python
grafo_condicional.add_conditional_edges(
    source="clasificar", # Nodo de origen
    path = router, # Funcion que decide el destino o siguiente nodo
    path_map={
        "matematicas": "matematicas",
        "cultura": "cultura",
        "codigo": "codigo",
    }
)
```

```python
grafo_condicional.add_edge("matematicas", END)
grafo_condicional.add_edge("cultura", END)
grafo_condicional.add_edge("codigo", END)
```

```python
app_condicional = grafo_condicional.compile()
```

```python
# Visualización con Mermaid (si estás en Jupyter/Colab)
try:
    from IPython.display import Image, display
    display(Image(app_condicional.get_graph().draw_mermaid_png()))
except Exception as e:
    print("Visualización gráfica no disponible:", e)
    print("Usa draw_ascii() en su lugar")
```

```python
def ejecutar_asistente(pregunta: str):
    """Helper para ejecutar el asistente y mostrar el resultado."""
    print("\n" + "=" * 65)
    print(f"PREGUNTA: {pregunta}")
    print("=" * 65)

    estado_inicial = {
        "pregunta": pregunta,
        "tipo": "",
        "respuesta": "",
        "nodo_usado": ""
    }

    resultado_condicional = app_condicional.invoke(estado_inicial)

    print(f"\n✅ Nodo usado: '{resultado_condicional['nodo_usado']}'")
    print(f"\nRESPUESTA:\n{resultado_condicional['respuesta']}")

    return resultado_condicional
```

```python
from typing import Annotated, TypedDict
from langgraph.graph.message import add_messages
from langchain_core.messages import BaseMessage
```

```python
class EstadoChat(TypedDict):
  mensajes: Annotated[list[BaseMessage], add_messages]
```

```python
from langchain_core.messages import SystemMessage
```

```python
system_prompt = """ Eres un asistente amigable que recuerda todo lo que el usuario te ha dicho en la conversacion.
Personaliza tus respuestas con la informacion que el usuario te ha compartido.
Si no sabes o no tienes en tu memoria lo que te está preguntando o te falta algo de informacion di que no lo sabes.
"""
```

```python
def nodo_chatbot(estado: EstadoChat) -> dict:
  """ Recibe el historial completo de mensajes y genera una respuesta.
  El historial incluye TODOS los turnos anteriores
  """

  mensajes_completos = [SystemMessage(content=system_prompt)] + estado["mensajes"]

  respuesta = llm.invoke(mensajes_completos)

  return {
      "mensajes": [respuesta]
  }
```

```python
from langgraph.checkpoint.memory import MemorySaver
```

```python
grafo_memoria = StateGraph(EstadoChat)
```

```python
grafo_memoria.add_node("chatbot", nodo_chatbot)
```

```python
grafo_memoria.add_edge(START, "chatbot")
grafo_memoria.add_edge("chatbot", END)
```

```python
memoria = MemorySaver()
```

```python
app_memoria = grafo_memoria.compile(checkpointer=memoria)
```

```python
from langchain_core.messages import HumanMessage
```

```python
def chatear(app, mensaje: str, thread_id: str):
    """Envía un mensaje al chatbot dentro de un thread específico."""
    # El config es OBLIGATORIO cuando usamos checkpointer
    config = {"configurable": {"thread_id": thread_id}}

    estado_entrada = {"mensajes": [HumanMessage(content=mensaje)]}

    resultado = app.invoke(estado_entrada, config=config)

    # El último mensaje es la respuesta del AI
    respuesta = resultado["mensajes"][-1].content
    return respuesta
```

# Herramientas

El patrón ReAct explicado

Pensamiento: "Necesito saber el clima de Madrid"
    ↓
Acción: llamar tool get_clima(ciudad="Madrid")
    ↓
Observación: "18°C, parcialmente nublado"
    ↓
Pensamiento: "Ahora sé el clima, puedo responder"
    ↓
Respuesta Final: "El clima en Madrid es 18°C..."

```python
from ast import Expression
from langchain_core.tools import tool
import math
import random

@tool
def calcular(expresion: str) -> str:
  """ Evalua una expresion matematica. Usa esto para cualquier calculo numerico.
  Soporta operaciones basicas (+,-,*,/), potencias (**) y funciones matematicas como sqrt,sin,con,log.
  Ejemplo: '2 ** 10', 'math.sqrt(144)', '15*7 +3'"""

  try:
      # Evaluamos de forma segura con math disponible
      resultado = eval(expresion, {"__builtins__": {}}, {"math": math, "sqrt": math.sqrt})
      return f"Resultado: {resultado}"
  except Exception as e:
      return f"Error al calcular '{expresion}': {e}"

@tool
def obtener_clima(ciudad: str) -> str:
    """Obtiene el clima actual de una ciudad.
    Retorna temperatura, condición y humedad.
    Úsalo cuando el usuario pregunte por el tiempo o clima en algún lugar.
    """
    # Simulamos una API de clima con datos ficticios
    climas_falsos = {
        "madrid": {"temp": 22, "condicion": "Soleado",          "humedad": 40},
        "barcelona": {"temp": 19, "condicion": "Parcialmente nublado", "humedad": 65},
        "sevilla": {"temp": 29, "condicion": "Muy caluroso",    "humedad": 30},
        "bilbao": {"temp": 15, "condicion": "Lluvioso",         "humedad": 85},
        "default": {"temp": random.randint(10, 30), "condicion": "Variable", "humedad": 60}
    }
    datos = climas_falsos.get(ciudad.lower(), climas_falsos["default"])
    return (f"Clima en {ciudad}: {datos['temp']}°C, {datos['condicion']}, "
            f"humedad {datos['humedad']}%")

@tool
def buscar_wikipedia(tema: str) -> str:
    """Busca información sobre un tema en Wikipedia.
    Retorna un resumen conciso. Úsalo cuando necesites información factual
    sobre personas, lugares, eventos históricos o conceptos.
    """
    # Simulamos Wikipedia con datos hardcoded
    base_datos = {
        "python": "Python es un lenguaje de programación de alto nivel creado por Guido van Rossum en 1991. Es conocido por su sintaxis clara y su enfoque en la legibilidad.",
        "madrid": "Madrid es la capital y ciudad más grande de España, con una población de más de 3 millones de habitantes. Es conocida por el Museo del Prado y el estadio Santiago Bernabéu.",
        "einstein": "Albert Einstein (1879–1955) fue un físico alemán-estadounidense que desarrolló la teoría de la relatividad. Ganó el Premio Nobel de Física en 1921.",
        "langgraph": "LangGraph es un framework de Python para construir agentes y flujos de trabajo con LLMs usando grafos de estado. Fue creado por LangChain Inc."
    }
    resultado = base_datos.get(tema.lower())
    if resultado:
        return resultado
    return f"Información sobre '{tema}': Tema encontrado con información general. Es un concepto/entidad relevante en su dominio."
```

```python
tools = [calcular, obtener_clima, buscar_wikipedia]
```

```python
llm_with_tools = llm.bind_tools(tools)
```

```python
class EstadoAgente(TypedDict):
  mensajes: Annotated[list[BaseMessage], add_messages]
```

```python
SYSTEM_REACT = """Eres un asistente inteligente con acceso a herramientas.
Razona paso a paso y usa las herramientas disponibles cuando sea necesario.
Para preguntas que requieran cálculos, usa la tool 'calcular'.
Para clima, usa 'obtener_clima'. Para información factual, usa 'buscar_wikipedia'."""
```

```python
def nodo_agente(estado: EstadoAgente) -> dict:
  """
  El nodo principal del agente. Llama al LLM y puede:
  1. Retornar tool_calls -> El grafo ejecuta las tools
  2. Retornar texto final -> El grafo termina
  """
  mensajes = [SystemMessage(content=SYSTEM_REACT)] + estado["mensajes"]

  respuesta = llm_with_tools.invoke(mensajes)

  if respuesta.tool_calls:
    print(f"[Agente] Solicitando {len(respuesta.tool_calls)} tool(s):")
    for tc in respuesta.tool_calls:
      print(f"  → {tc['name']}({tc['args']})")
  else:
    print("[Agente] Generando respuesta final (sin mas tools)")

  return {
      "mensajes": [respuesta]
  }
```

```python
from langgraph.prebuilt import ToolNode
```

```python
nodo_tools = ToolNode(tools)
```

```python
def router_react(estado: EstadoAgente) -> Literal["tools", "__end__"]:
  """ Si el ultimo mensaje del agente tiene tool_calls -> ejecutar tools
  Si no -> terminar (respuesta final generada)
  """
  ultimo_mensaje = estado["mensajes"][-1]

  if hasattr(ultimo_mensaje, "tool_calls") and ultimo_mensaje.tool_calls:
    return "tools"    # Continúa el ciclo

  return "__end__"      # Termina el grafo
```

```python
grafo_react = StateGraph(EstadoAgente)

# Nodos
grafo_react.add_node("agente", nodo_agente)
grafo_react.add_node("tools", nodo_tools)

# Edges
grafo_react.add_edge(START, "agente")

# Condicional
grafo_react.add_conditional_edges(
    "agente",
    router_react,
    {"tools": "tools", "__end__": END}
)

grafo_react.add_edge("tools", "agente")
```

```python
app_react = grafo_react.compile(checkpointer=MemorySaver())
```

```python
# Visualización con Mermaid (si estás en Jupyter/Colab)
try:
    from IPython.display import Image, display
    display(Image(app_react.get_graph().draw_mermaid_png()))
except Exception as e:
    print("Visualización gráfica no disponible:", e)
    print("Usa draw_ascii() en su lugar")
```
